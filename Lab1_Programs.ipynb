```python
{
  "nbformat": 4,
  "nbformat_minor": 5,
  "metadata": {
    "kernelspec": {
      "display_name": "Python 3",
      "language": "python",
      "name": "python3"
    },
    "language_info": {
      "name": "python",
      "pygments_lexer": "ipython3"
    },
    "colab": {
      "provenance": []
    }
  },
  "cells": [
    {
      "cell_type": "markdown",
      "metadata": {
        "id": "S-o50orfoGOB"
      },
      "source": [
        "# Artificial Intelligence Lab 1\n",
        "\n",
        "Small Python programs for Google Colab. Run each code cell with Shift+Enter. When a program asks for input, type your answer in the box under the cell.\n"
      ],
      "id": "S-o50orfoGOB"
    },
    {
      "cell_type": "markdown",
      "metadata": {
        "id": "g6xAEUydoGOF"
      },
      "source": [
        "## Program 1: Hello World and user input\n",
        "\n",
        "Prints a greeting, then asks for your name.\n"
      ],
      "id": "g6xAEUydoGOF"
    },
    {
      "cell_type": "code",
      "execution_count": null,
      "metadata": {
        "id": "r5zVqMFMoGOG"
      },
      "outputs": [],
      "source": [
        "\"\"\"Lab 1 - Program 1: Hello World and user input.\"\"\"\n",
        "\n",
        "print(\"Hello, World!\")\n",
        "print(\"Welcome to Artificial Intelligence Lab 1\")\n",
        "\n",
        "name = input(\"Enter your name: \")\n",
        "print(f\"Hello, {name}! This program is running in Python.\")\n"
      ],
      "id": "r5zVqMFMoGOG"
    },
    {
      "cell_type": "code",
      "source": [
        "from collections import deque\n",
        "import heapq\n",
        "\n",
        "# ---------- Grid maze: S = start, G = goal, # = wall ----------\n",
        "MAZE = [\"S . . . . # #\",\n",
        "        \"# # . # # . .\",\n",
        "        \". . . # . . #\",\n",
        "        \". # . . . . #\",\n",
        "        \". . . . # . G\"]\n",
        "grid = [row.split() for row in MAZE]\n",
        "ROWS, COLS = len(grid), len(grid[0])\n",
        "\n",
        "def find(ch):\n",
        "    for r in range(ROWS):\n",
        "        for c in range(COLS):\n",
        "            if grid[r][c] == ch:\n",
        "                return (r, c)\n",
        "\n",
        "def neighbours(cell):\n",
        "    r, c = cell\n",
        "    for dr, dc in [(-1, 0), (0, 1), (1, 0), (0, -1)]:   # Up, Right, Down, Left\n",
        "        nr, nc = r + dr, c + dc\n",
        "        if 0 <= nr < ROWS and 0 <= nc < COLS and grid[nr][nc] != '#':\n",
        "            yield (nr, nc)\n",
        "\n",
        "def bfs(start, goal):\n",
        "    frontier = deque([start]); parent = {start: None}; expanded = 0\n",
        "    while frontier:\n",
        "        node = frontier.popleft()                 # FIFO queue\n",
        "        expanded += 1\n",
        "        if node == goal:\n",
        "            return build_path(parent, goal), expanded\n",
        "        for nb in neighbours(node):\n",
        "            if nb not in parent:\n",
        "                parent[nb] = node\n",
        "                frontier.append(nb)\n",
        "    return None, expanded\n",
        "\n",
        "def dfs(start, goal):\n",
        "    frontier = [start]; parent = {start: None}; expanded = 0\n",
        "    visited = set()\n",
        "    while frontier:\n",
        "        node = frontier.pop()                     # LIFO stack\n",
        "        if node in visited:\n",
        "            continue\n",
        "        visited.add(node); expanded += 1\n",
        "        if node == goal:\n",
        "            return build_path(parent, goal), expanded\n",
        "        for nb in neighbours(node):\n",
        "            if nb not in visited:\n",
        "                parent[nb] = node\n",
        "                frontier.append(nb)\n",
        "    return None, expanded\n",
        "\n",
        "def build_path(parent, node):\n",
        "    path = []\n",
        "    while node is not None:\n",
        "        path.append(node); node = parent[node]\n",
        "    return path[::-1]\n",
        "\n",
        "S, G = find('S'), find('G')\n",
        "print(f\"{'Algorithm':<6}{'Path length':>13}{'Nodes expanded':>17}\")\n",
        "for name, algo in [('BFS', bfs), ('DFS', dfs)]:\n",
        "    path, exp = algo(S, G)\n",
        "    print(f\"{name:<6}{len(path)-1:>13}{exp:>17}\")\n",
        "\n",
        "# ---------- Uniform Cost Search on a weighted graph ----------\n",
        "graph = {\n",
        "    'Islamabad':  {'Rawalpindi': 15, 'Murree': 60, 'Attock': 80},\n",
        "    'Rawalpindi': {'Islamabad': 15, 'Chakwal': 90, 'Jhelum': 110},\n",
        "    'Murree':     {'Islamabad': 60, 'Abbottabad': 65},\n",
        "    'Attock':     {'Islamabad': 80, 'Peshawar': 100},\n",
        "    'Chakwal':    {'Rawalpindi': 90, 'Lahore': 230},\n",
        "    'Jhelum':     {'Rawalpindi': 110, 'Gujranwala': 110},\n",
        "    'Gujranwala': {'Jhelum': 110, 'Lahore': 70},\n",
        "    'Abbottabad': {'Murree': 65},\n",
        "    'Peshawar':   {'Attock': 100},\n",
        "    'Lahore':     {'Chakwal': 230, 'Gujranwala': 70},\n",
        "}\n",
        "\n",
        "def ucs(graph, start, goal):\n",
        "    frontier = [(0, start, [start])]          # (path cost, node, path)\n",
        "    explored = set(); expanded = 0\n",
        "    while frontier:\n",
        "        cost, node, path = heapq.heappop(frontier)   # lowest g(n) first\n",
        "        if node in explored:\n",
        "            continue\n",
        "        explored.add(node); expanded += 1\n",
        "        if node == goal:\n",
        "            return path, cost, expanded\n",
        "        for nb, w in graph[node].items():\n",
        "            if nb not in explored:\n",
        "                heapq.heappush(frontier, (cost + w, nb, path + [nb]))\n",
        "    return None, float('inf'), expanded\n",
        "\n",
        "path, cost, exp = ucs(graph, 'Islamabad', 'Lahore')\n",
        "print(\"\\nUCS path :\", ' -> '.join(path))\n",
        "print(\"UCS cost :\", cost, \"km | nodes expanded:\", exp)"
      ],
      "metadata": {
        "id": "v3mMog5coLiZ",
        "outputId": "0e720479-21e8-44a1-a76a-cf982d149f2e",
        "colab": {
          "base_uri": "https://localhost:8080/"
        }
      },
      "id": "v3mMog5coLiZ",
      "execution_count": 3,
      "outputs": [
        {
          "output_type": "stream",
          "name": "stdout",
          "text": [
            "Algorithm  Path length   Nodes expanded\n",
            "BFS              10               23\n",
            "DFS              16               18\n",
            "\n",
            "UCS path : Islamabad -> Rawalpindi -> Jhelum -> Gujranwala -> Lahore\n",
            "UCS cost : 305 km | nodes expanded: 10\n"
          ]
        }
      ]
    },
    {
      "cell_type": "markdown",
      "metadata": {
        "id": "Fcln-cRyoGOH"
      },
      "source": [
        "## Program 2: Variables and data types\n",
        "\n",
        "Shows string, int, float, and bool, then converts marks from text to a number.\n"
      ],
      "id": "Fcln-cRyoGOH"
    },
    {
      "cell_type": "code",
      "execution_count": null,
      "metadata": {
        "id": "MeGzrYM_oGOI"
      },
      "outputs": [],
      "source": [
        "\"\"\"Lab 1 - Program 2: Variables and data types.\"\"\"\n",
        "\n",
        "course = \"Artificial Intelligence\"\n",
        "lab_number = 1\n",
        "score = 92.5\n",
        "is_completed = False\n",
        "\n",
        "print(\"Course:\", course, \"| type:\", type(course).__name__)\n",
        "print(\"Lab number:\", lab_number, \"| type:\", type(lab_number).__name__)\n",
        "print(\"Score:\", score, \"| type:\", type(score).__name__)\n",
        "print(\"Completed:\", is_completed, \"| type:\", type(is_completed).__name__)\n",
        "\n",
        "marks_text = input(\"Enter your marks (0-100): \")\n",
        "marks = float(marks_text)\n",
        "print(\"Marks as float:\", marks)\n",
        "print(\"Marks as integer:\", int(marks))\n",
        "print(\"Passed:\", marks >= 50)\n"
      ],
      "id": "MeGzrYM_oGOI"
    },
    {
      "cell_type": "markdown",
      "metadata": {
        "id": "Kl4ad-PioGOI"
      },
      "source": [
        "## Program 3: Arithmetic calculator\n",
        "\n",
        "Adds, subtracts, multiplies, or divides two numbers.\n"
      ],
      "id": "Kl4ad-PioGOI"
    },
    {
      "cell_type": "code",
      "execution_count": null,
      "metadata": {
        "id": "k3Pc1FPZoGOJ"
      },
      "outputs": [],
      "source": [
        "\"\"\"Lab 1 - Program 3: Simple arithmetic calculator.\"\"\"\n",
        "\n",
        "print(\"Simple Calculator\")\n",
        "print(\"Operations: +  -  *  /\")\n",
        "\n",
        "first = float(input(\"Enter the first number: \"))\n",
        "operator = input(\"Enter an operator (+, -, *, /): \").strip()\n",
        "second = float(input(\"Enter the second number: \"))\n",
        "\n",
        "if operator == \"+\":\n",
        "    result = first + second\n",
        "elif operator == \"-\":\n",
        "    result = first - second\n",
        "elif operator == \"*\":\n",
        "    result = first * second\n",
        "elif operator == \"/\":\n",
        "    if second == 0:\n",
        "        print(\"Cannot divide by zero.\")\n",
        "        result = None\n",
        "    else:\n",
        "        result = first / second\n",
        "else:\n",
        "    print(\"Unknown operator.\")\n",
        "    result = None\n",
        "\n",
        "if result is not None:\n",
        "    print(f\"{first} {operator} {second} = {result}\")\n"
      ],
      "id": "k3Pc1FPZoGOJ"
    },
    {
      "cell_type": "markdown",
      "metadata": {
        "id": "wfqeTowVoGOJ"
      },
      "source": [
        "## Program 4: Conditions\n",
        "\n",
        "Checks even or odd, positive or negative, and assigns a letter grade.\n"
      ],
      "id": "wfqeTowVoGOJ"
    },
    {
      "cell_type": "code",
      "execution_count": null,
      "metadata": {
        "id": "WtYt5SREoGOK"
      },
      "outputs": [],
      "source": [
        "\"\"\"Lab 1 - Program 4: Conditions (even/odd and letter grade).\"\"\"\n",
        "\n",
        "number = int(input(\"Enter an integer: \"))\n",
        "\n",
        "if number % 2 == 0:\n",
        "    print(f\"{number} is even.\")\n",
        "else:\n",
        "    print(f\"{number} is odd.\")\n",
        "\n",
        "if number > 0:\n",
        "    print(\"The number is positive.\")\n",
        "elif number < 0:\n",
        "    print(\"The number is negative.\")\n",
        "else:\n",
        "    print(\"The number is zero.\")\n",
        "\n",
        "marks = float(input(\"Enter marks out of 100: \"))\n",
        "\n",
        "if marks >= 90:\n",
        "    grade = \"A\"\n",
        "elif marks >= 80:\n",
        "    grade = \"B\"\n",
        "elif marks >= 70:\n",
        "    grade = \"C\"\n",
        "elif marks >= 60:\n",
        "    grade = \"D\"\n",
        "elif marks >= 50:\n",
        "    grade = \"E\"\n",
        "else:\n",
        "    grade = \"F\"\n",
        "\n",
        "print(f\"Grade: {grade}\")\n"
      ],
      "id": "WtYt5SREoGOK"
    },
    {
      "cell_type": "markdown",
      "metadata": {
        "id": "mEnwe43MoGOK"
      },
      "source": [
        "## Program 5: Loops\n",
        "\n",
        "Prints a multiplication table, a running sum, and a factorial.\n"
      ],
      "id": "mEnwe43MoGOK"
    },
    {
      "cell_type": "code",
      "execution_count": null,
      "metadata": {
        "id": "ioqe86NSoGOK"
      },
      "outputs": [],
      "source": [
        "\"\"\"Lab 1 - Program 5: Loops (multiplication table, sum, and factorial).\"\"\"\n",
        "\n",
        "number = int(input(\"Enter a positive integer: \"))\n",
        "\n",
        "print(f\"\\nMultiplication table of {number}\")\n",
        "for i in range(1, 11):\n",
        "    print(f\"{number} x {i} = {number * i}\")\n",
        "\n",
        "total = 0\n",
        "count = 1\n",
        "while count <= number:\n",
        "    total += count\n",
        "    count += 1\n",
        "print(f\"\\nSum of numbers from 1 to {number} = {total}\")\n",
        "\n",
        "factorial = 1\n",
        "for i in range(1, number + 1):\n",
        "    factorial *= i\n",
        "print(f\"Factorial of {number} = {factorial}\")\n"
      ],
      "id": "ioqe86NSoGOK"
    },
    {
      "cell_type": "markdown",
      "metadata": {
        "id": "IabLAoxpoGOL"
      },
      "source": [
        "## Program 6: Lists\n",
        "\n",
        "Reads several numbers into a list and prints count, sum, average, min, and max.\n"
      ],
      "id": "IabLAoxpoGOL"
    },
    {
      "cell_type": "code",
      "execution_count": null,
      "metadata": {
        "id": "LgnqTnsooGOL"
      },
      "outputs": [],
      "source": [
        "\"\"\"Lab 1 - Program 6: Lists (store numbers, then summarize them).\"\"\"\n",
        "\n",
        "raw = input(\"Enter numbers separated by spaces: \")\n",
        "numbers = [float(item) for item in raw.split()]\n",
        "\n",
        "if not numbers:\n",
        "    print(\"No numbers were entered.\")\n",
        "else:\n",
        "    print(\"Numbers:\", numbers)\n",
        "    print(\"Count:\", len(numbers))\n",
        "    print(\"Sum:\", sum(numbers))\n",
        "    print(\"Average:\", sum(numbers) / len(numbers))\n",
        "    print(\"Minimum:\", min(numbers))\n",
        "    print(\"Maximum:\", max(numbers))\n",
        "    print(\"Sorted:\", sorted(numbers))\n"
      ],
      "id": "LgnqTnsooGOL"
    },
    {
      "cell_type": "markdown",
      "metadata": {
        "id": "eEf6FkDQoGOL"
      },
      "source": [
        "## Program 7: Functions\n",
        "\n",
        "Defines functions for area, prime checking, and temperature conversion.\n"
      ],
      "id": "eEf6FkDQoGOL"
    },
    {
      "cell_type": "code",
      "execution_count": null,
      "metadata": {
        "id": "jS7CgWYmoGOL"
      },
      "outputs": [],
      "source": [
        "\"\"\"Lab 1 - Program 7: Functions.\"\"\"\n",
        "\n",
        "\n",
        "def rectangle_area(length, width):\n",
        "    return length * width\n",
        "\n",
        "\n",
        "def is_prime(value):\n",
        "    if value < 2:\n",
        "        return False\n",
        "    for divisor in range(2, int(value ** 0.5) + 1):\n",
        "        if value % divisor == 0:\n",
        "            return False\n",
        "    return True\n",
        "\n",
        "\n",
        "def celsius_to_fahrenheit(celsius):\n",
        "    return (celsius * 9 / 5) + 32\n",
        "\n",
        "\n",
        "length = float(input(\"Enter rectangle length: \"))\n",
        "width = float(input(\"Enter rectangle width: \"))\n",
        "print(\"Area:\", rectangle_area(length, width))\n",
        "\n",
        "candidate = int(input(\"Enter an integer to test for prime: \"))\n",
        "if is_prime(candidate):\n",
        "    print(f\"{candidate} is prime.\")\n",
        "else:\n",
        "    print(f\"{candidate} is not prime.\")\n",
        "\n",
        "celsius = float(input(\"Enter temperature in Celsius: \"))\n",
        "print(f\"{celsius} C = {celsius_to_fahrenheit(celsius)} F\")\n"
      ],
      "id": "jS7CgWYmoGOL"
    },
    {
      "cell_type": "markdown",
      "metadata": {
        "id": "ohiNk3RToGOL"
      },
      "source": [
        "## Program 8: Rule-based chatbot\n",
        "\n",
        "Replies to a few keywords. Type quit to stop.\n"
      ],
      "id": "ohiNk3RToGOL"
    },
    {
      "cell_type": "code",
      "execution_count": null,
      "metadata": {
        "id": "es2EzQUHoGOM"
      },
      "outputs": [],
      "source": [
        "\"\"\"Lab 1 - Program 8: A small rule-based chatbot.\"\"\"\n",
        "\n",
        "print(\"AI Lab Chatbot\")\n",
        "print(\"Type a message, or type 'quit' to stop.\")\n",
        "print(\"Try: hello, your name, python, ai, bye\")\n",
        "\n",
        "replies = {\n",
        "    \"hello\": \"Hello! I am a small rule-based chatbot.\",\n",
        "    \"hi\": \"Hi! How can I help you with Lab 1?\",\n",
        "    \"your name\": \"My name is LabBot. I only know a few replies.\",\n",
        "    \"python\": \"Python is the language we are using in Google Colab.\",\n",
        "    \"ai\": \"Artificial intelligence lets computers learn patterns and make decisions.\",\n",
        "    \"bye\": \"Goodbye! Good luck with your lab.\",\n",
        "}\n",
        "\n",
        "while True:\n",
        "    message = input(\"You: \").strip().lower()\n",
        "    if message in (\"quit\", \"exit\", \"bye\"):\n",
        "        print(\"LabBot: Goodbye!\")\n",
        "        break\n",
        "    if message in replies:\n",
        "        print(\"LabBot:\", replies[message])\n",
        "    else:\n",
        "        print(\"LabBot: I do not know that yet. Try hello, python, or ai.\")\n"
      ],
      "id": "es2EzQUHoGOM"
    }
  ]
}
```
